```python
# 148 Write a program to calculate the area of different shapes using OOP.

class circle:
    def area(self,r):
        return 3.14 * r * r

class rectangle:
    def area(self,l,b):
        return l * b

c = circle()
r = rectangle()

print("circle area:",c.area(5))
print("rectangle area:",r.area(4,6))
```

```python
# 161 Write a function that filters rows in a DataFrame where two conditions are met, e.g., a column value is greater than a number, 
#and another column contains a specific string.

import pandas as pd

data = {
    "Name": ["Ram", "Shyam", "Rahul"],
    "Marks": [85, 65, 90]
}

df = pd.DataFrame(data)

result = df[(df["Marks"] > 70) & (df["Name"].str.contains("R"))]

print(result)
```
